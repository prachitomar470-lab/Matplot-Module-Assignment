```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

users = [1200, 1400, 1350, 1600, 1800, 2000,
         2200, 2100, 2400, 2600, 2500, 3000]

plt.plot(months, users, marker="o")

highest = max(users)
index = users.index(highest)

plt.annotate("Highest: " + str(highest),
             xy=(months[index], highest),
             xytext=(months[index-2], highest-300),
             arrowprops=dict(arrowstyle="->", color="red"))

plt.title("Monthly Active Users")
plt.xlabel("Months")
plt.ylabel("Number of Users")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

orders = [450, 500, 380, 600, 550, 700,
          650, 480, 750, 800, 720, 900]

plt.bar(months, orders, color="orange")

lowest = min(orders)
index = orders.index(lowest)

plt.annotate("Lowest: " + str(lowest),
             xy=(index, lowest),
             xytext=(index + 1, lowest + 150),
             arrowprops=dict(arrowstyle="->", color="red"))

plt.title("Monthly Zomato Orders")
plt.xlabel("Months")
plt.ylabel("Number of Orders")

plt.show()
```

```python
import matplotlib.pyplot as plt

methods = ["UPI", "Card", "Wallet", "Cash"]
payments = [50, 25, 15, 10]

plt.pie(payments, labels=methods, autopct="%1.1f%%")

plt.title("E-commerce Payment Methods")
plt.legend(methods, loc="lower right")

plt.show()
```

```python
import matplotlib.pyplot as plt

movies = ["Movie A", "Movie B", "Movie C", "Movie D", "Movie E"]
ratings = [7.5, 8.2, 9.1, 7.8, 8.5]
reviews = [200, 350, 500, 250, 400]

plt.scatter(reviews, ratings)

highest = max(ratings)
index = ratings.index(highest)

plt.annotate(movies[index] + ": " + str(highest),
             xy=(reviews[index], highest),
             xytext=(reviews[index] + 30, highest - 0.5),
             bbox=dict(boxstyle="round", facecolor="yellow", edgecolor="black"),
             arrowprops=dict(arrowstyle="->", color="red"))

plt.title("Movie Ratings vs Number of Reviews")
plt.xlabel("Number of Reviews")
plt.ylabel("Movie Rating")

plt.show()
```
