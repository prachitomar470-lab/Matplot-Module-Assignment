```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4000, 5000, 6500, 7000, 6000, 9000, 8000]
water = [2, 2.5, 2, 3, 2.5, 3, 2.5]

plt.figure(figsize=(8, 6))

plt.subplot(2, 1, 1)
plt.plot(days, steps, marker="o")
plt.title("Daily Step Count")
plt.xlabel("Days")
plt.ylabel("Steps")

plt.subplot(2, 1, 2)
plt.plot(days, water, marker="o", color="green")
plt.title("Daily Water Intake")
plt.xlabel("Days")
plt.ylabel("Water (Litres)")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [20, 30, 25, 40]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# Line chart
ax[0, 0].plot(quarters, posts, marker="o")
ax[0, 0].set_title("Line Chart")

# Bar chart
ax[0, 1].bar(quarters, posts)
ax[0, 1].set_title("Bar Chart")

# Scatter plot
ax[1, 0].scatter(quarters, posts)
ax[1, 0].set_title("Scatter Plot")

# Pie chart
ax[1, 1].pie(posts, labels=quarters, autopct="%1.1f%%")
ax[1, 1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

zomato = [4, 5, 3, 6, 7, 5]
swiggy = [3, 4, 5, 4, 6, 7]
dominos = [2, 3, 4, 3, 5, 4]

fig, ax = plt.subplots(1, 3, figsize=(12, 4), sharey=True)

ax[0].plot(months, zomato, marker="o")
ax[0].set_title("Zomato Orders")
ax[0].set_xlabel("Months")
ax[0].set_ylabel("Number of Orders")

ax[1].plot(months, swiggy, marker="o", color="green")
ax[1].set_title("Swiggy Orders")
ax[1].set_xlabel("Months")

ax[2].plot(months, dominos, marker="o", color="red")
ax[2].set_title("Domino's Orders")
ax[2].set_xlabel("Months")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

youtube = [8, 10, 12, 15, 14, 18]
instagram = [4, 5, 6, 7, 8, 9]
spotify = [2, 3, 2, 4, 3, 5]
whatsapp = [1, 2, 2, 3, 2, 3]

fig, ax = plt.subplots(2, 2, figsize=(10, 8), sharex=True)

ax[0, 0].plot(months, youtube, marker="o")
ax[0, 0].set_title("YouTube Data Usage")
ax[0, 0].set_ylabel("Data Usage (GB)")

ax[0, 1].plot(months, instagram, marker="o", color="orange")
ax[0, 1].set_title("Instagram Data Usage")
ax[0, 1].set_ylabel("Data Usage (GB)")

ax[1, 0].plot(months, spotify, marker="o", color="green")
ax[1, 0].set_title("Spotify Data Usage")
ax[1, 0].set_xlabel("Months")
ax[1, 0].set_ylabel("Data Usage (GB)")

ax[1, 1].plot(months, whatsapp, marker="o", color="red")
ax[1, 1].set_title("WhatsApp Data Usage")
ax[1, 1].set_xlabel("Months")
ax[1, 1].set_ylabel("Data Usage (GB)")

plt.tight_layout()
plt.show()
```
