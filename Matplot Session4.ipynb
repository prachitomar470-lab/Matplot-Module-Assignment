```python

import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]
hours = [3, 2, 2, 1, 2]

plt.pie(hours, labels=apps, autopct="%1.1f%%")
plt.show()

```

```python

import matplotlib.pyplot as plt

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [1000, 3000, 500, 1500]
explode = [0, 0.1, 0, 0]

plt.pie(spending, labels=categories, explode=explode, autopct="%1.1f%%")
plt.show()
```

```python

import matplotlib.pyplot as plt

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [1000, 3000, 500, 1500]

fig, ax = plt.subplots()

ax.pie(spending, labels=categories, autopct="%1.1f%%")

circle = plt.Circle((0, 0), 0.5, color="white")
ax.add_artist(circle)

plt.show()
```

```python

import matplotlib.pyplot as plt

teams = ["CSK", "MI", "RCB", "KKR"]
followers = [15, 14, 12, 8]

fig, ax = plt.subplots()

ax.pie(followers, labels=teams, autopct="%1.1f%%")

circle = plt.Circle((0, 0), 0.5, color="white")
ax.add_artist(circle)

plt.title("IPL Teams Instagram Followers Share")
plt.show()

```
