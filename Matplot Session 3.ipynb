```python

import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Dominos"]
orders = [100, 150, 80, 60, 120]

plt.bar(apps, orders, color=["red", "blue", "green", "orange", "purple"])
plt.show()

```

```python

import matplotlib.pyplot as plt

names = ["A", "B", "C", "D", "E", "F"]
followers = [2, 5, 3, 7, 4, 6]

plt.barh(names, followers)
plt.title("Instagram Followers")
plt.xlabel("Followers (in millions)")
plt.show()

```

```python

import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Dominos"]
orders = [100, 150, 80, 60, 120]

plt.bar(apps, orders)

for i in range(len(orders)):
    plt.text(i, orders[i], orders[i])

plt.show()

```
