```python
import pandas as pd

data = {
    "Product": ["Laptop", "T-Shirt", "Rice", "Mobile", "Jeans",
                "Headphones", "Shirt", "Wheat", "Tablet", "Shoes",
                "Laptop", "Rice", "Mobile", "Jeans", "Headphones",
                "T-Shirt", "Wheat", "Tablet", "Shoes", "Shirt"],

    "Category": ["Electronics", "Clothing", "Groceries", "Electronics", "Clothing",
                 "Electronics", "Clothing", "Groceries", "Electronics", "Clothing",
                 "Electronics", "Groceries", "Electronics", "Clothing", "Electronics",
                 "Clothing", "Groceries", "Electronics", "Clothing", "Clothing"],

    "Date": ["2026-01-05", "2026-01-12", "2026-02-10", "2026-02-15", "2026-03-08",
             "2026-03-20", "2026-04-05", "2026-04-18", "2026-05-10", "2026-05-22",
             "2026-06-12", "2026-07-03", "2026-07-15", "2026-08-08", "2026-09-11",
             "2026-10-04", "2026-10-18", "2026-11-06", "2026-12-02", "2026-12-15"],

    "Sales": [50000, 1200, 2500, 30000, 1800,
              2500, 1500, 2200, 20000, 3000,
              55000, 2800, 35000, 2200, 3000,
              1400, 2600, 18000, 3200, 1700]
}

df = pd.DataFrame(data)

df["Date"] = pd.to_datetime(df["Date"])

df.to_csv("sales_dataset.csv", index=False)

df.head()
```

```python
import matplotlib.pyplot as plt

category_sales = df.groupby("Category")["Sales"].sum()

plt.bar(category_sales.index, category_sales.values, color="skyblue")

plt.title("Total Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales (Rs.)")

plt.show()
```

```python
monthly_sales = df.groupby(df["Date"].dt.month)["Sales"].sum()

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

plt.figure(figsize=(10, 5))

plt.plot(months, monthly_sales.values, marker="o", color="green")

plt.title("Monthly Sales Trend - 2026")
plt.xlabel("Month")
plt.ylabel("Total Sales (Rs.)")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=10, color="orange", edgecolor="black")

plt.title("Distribution of Order Amounts")
plt.xlabel("Sales Amount (Rs.)")
plt.ylabel("Number of Orders")

plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(8, 5))

bars = plt.bar(category_sales.index, category_sales.values, color="skyblue")

plt.title("Total Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales (Rs.)")

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(height),
        ha="center",
        va="bottom"
    )

plt.savefig("category_sales_kpi.png", dpi=150, bbox_inches="tight")

plt.show()
```
