```python

import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
steps = [4500, 6000, 7500, 5000, 8000, 10000, 9000]

plt.plot(days, steps, marker="o")

plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.title("Daily Steps Over 7 Days")

plt.show()
```

```python

import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

spotify_hours = [1, 2, 1.5, 2.5, 3, 4, 2]
youtube_hours = [1.5, 1, 2, 1.5, 2, 3, 2.5]

fig, ax = plt.subplots(1, 2)

ax[0].plot(days, spotify_hours, marker="o")
ax[0].set_title("Spotify Listening Hours")

ax[1].plot(days, youtube_hours, marker="o")
ax[1].set_title("YouTube Viewing Hours")

plt.tight_layout()
plt.show()

```

```python

import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

spotify_hours = [1, 2, 1.5, 2.5, 3, 4, 2]
youtube_hours = [1.5, 1, 2, 1.5, 2, 3, 2.5]

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(days, spotify_hours, marker="o")
ax[0].set_xlabel("Days of the Week")
ax[0].set_ylabel("Listening Hours")
ax[0].set_title("Daily Spotify Listening Hours")
ax[0].grid(True)

ax[1].plot(days, youtube_hours, marker="o")
ax[1].set_xlabel("Days of the Week")
ax[1].set_ylabel("Viewing Hours")
ax[1].set_title("Daily YouTube Viewing Hours")
ax[1].grid(True)

plt.tight_layout()
plt.show()

```
