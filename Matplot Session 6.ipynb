```python

import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 2, 5, 3, 6, 4, 5]
posts = [5, 8, 10, 12, 7, 15, 9, 18, 13, 16]

plt.scatter(hours, posts)
plt.xlabel("Hours on Instagram")
plt.ylabel("Monthly Posts")
plt.title("Instagram Usage vs Posts")
plt.show()
```

```python

import matplotlib.pyplot as plt

names = ["A", "B", "C", "D", "E", "F",
         "G", "H", "I", "J", "K", "L"]

orders = [2, 4, 3, 5, 6, 2, 7, 4, 8, 5, 3, 6]
spending = [300, 600, 450, 800, 900, 250,
            1100, 650, 1300, 750, 400, 950]

plt.scatter(orders, spending)

for i in range(len(names)):
    plt.annotate(names[i], (orders[i], spending[i]))

plt.xlabel("Number of Orders")
plt.ylabel("Amount Spent (Rs.)")
plt.title("Zomato Orders vs Spending")
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

orders = [2, 4, 3, 5, 6, 2, 7, 4, 8, 5, 3, 6]
spending = [300, 600, 450, 800, 900, 250,
            1100, 650, 1300, 750, 400, 950]

m, b = np.polyfit(orders, spending, 1)

plt.scatter(orders, spending)
plt.plot(orders, np.array(orders) * m + b, color="red")

plt.xlabel("Number of Orders")
plt.ylabel("Amount Spent (Rs.)")
plt.title("Zomato Spending Trend")
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

youtube = [1, 2, 3, 4, 5, 2, 6, 3, 4, 5, 1, 7, 6, 3, 5]
spotify = [2, 3, 4, 5, 6, 3, 7, 4, 5, 6, 2, 8, 7, 4, 6]

m, b = np.polyfit(youtube, spotify, 1)

plt.scatter(youtube, spotify)
plt.plot(youtube, np.array(youtube) * m + b, color="red")

plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")
plt.title("YouTube vs Spotify Usage")
plt.text(1, 7, "y = " + str(round(m, 2)) + "x + " + str(round(b, 2)))

plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

orders = [1, 2, 3, 4, 5, 2, 6, 3, 7, 5]
spending = [500, 1000, 1500, 2000, 2500,
            900, 3000, 1400, 3500, 2400]

m, b = np.polyfit(orders, spending, 1)

plt.scatter(orders, spending)
plt.plot(orders, np.array(orders) * m + b, color="red")

plt.xlabel("Number of Flipkart Orders")
plt.ylabel("Total Amount Spent (Rs.)")
plt.title("Flipkart Orders vs Spending")
plt.show()
```
